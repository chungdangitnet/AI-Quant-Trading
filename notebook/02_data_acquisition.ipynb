```python
# Cell 1: Import mô-đun và Khởi tạo
import sys
import os
sys.path.append(os.path.abspath(".."))

from utils.data_fetcher import StockDataFetcher

fetcher = StockDataFetcher()

# Cell 2: Tải dữ liệu Cổ phiếu Mỹ (S&P 500 / AAPL) từ Yahoo Finance
df_aapl = fetcher.fetch_yahoo(ticker="AAPL", start_date="2020-01-01", end_date="2026-01-01")
fetcher.save_to_parquet(df_aapl, ticker="AAPL")
fetcher.save_to_sqlite(df_aapl, table_name="AAPL")

# Cell 3: Tải dữ liệu Crypto (BTCUSDT) từ Binance
df_btc = fetcher.fetch_binance(symbol="BTCUSDT", interval="1d", limit=1000)
fetcher.save_to_parquet(df_btc, ticker="BTCUSDT")
fetcher.save_to_sqlite(df_btc, table_name="BTCUSDT")

# Cell 4: Tải dữ liệu VNINDEX
df_vnindex = fetcher.fetch_vnindex_sample(ticker="VNINDEX")
fetcher.save_to_sqlite(df_vnindex, table_name="VNINDEX")

# Cell 5: Kiểm tra đọc lại dữ liệu từ SQLite
df_check = fetcher.load_from_sqlite("AAPL")
print("=== DỮ LIỆU AAPL ĐỌC TỪ SQLITE ===")
print(df_check.head())
print(f"Tổng số hàng: {len(df_check)}")
```
