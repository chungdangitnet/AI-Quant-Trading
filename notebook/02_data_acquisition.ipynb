```python
# Cell 1: Import mô-đun và Khởi tạo
import sys
import os
sys.path.append(os.path.abspath(".."))

from utils.data_fetcher import StockDataFetcher

fetcher = StockDataFetcher()

# Cell 2: Tải dữ liệu GAS từ Yahoo Finance
df_gas = fetcher.fetch_yahoo(ticker="GAS.VN", start_date="2020-01-01", end_date="2026-01-01")
fetcher.save_to_parquet(df_gas, ticker="GAS")
fetcher.save_to_sqlite(df_gas, table_name="GAS")

# Cell 3: Tải dữ liệu Crypto (BTCUSDT) từ Binance
df_btc = fetcher.fetch_binance(symbol="BTCUSDT", interval="1d", limit=1000)
fetcher.save_to_parquet(df_btc, ticker="BTCUSDT")
fetcher.save_to_sqlite(df_btc, table_name="BTCUSDT")

# Cell 4: Tải dữ liệu đại diện thị trường VN (Dùng ETF E1VFVN30.VN thay cho VNINDEX)
# Quỹ này mô phỏng sát diễn biến của VN-Index và được Yahoo hỗ trợ ổn định
df_vnindex = fetcher.fetch_yahoo(ticker="E1VFVN30.VN", start_date="2020-01-01", end_date="2026-01-01")

# Vẫn lưu tên bảng và file là VNINDEX để các bước sau không bị lỗi logic
fetcher.save_to_parquet(df_vnindex, ticker="VNINDEX")
fetcher.save_to_sqlite(df_vnindex, table_name="VNINDEX")

# Kiểm tra dữ liệu xem đã tải thành công chưa
print("=== DỮ LIỆU ĐẠI DIỆN VNINDEX (E1VFVN30) ===")
print(df_vnindex.head())

# Cell 5: Kiểm tra đọc lại dữ liệu từ SQLite
df_check = fetcher.load_from_sqlite("GAS")
print("=== DỮ LIỆU GAS ĐỌC TỪ SQLITE ===")
print(df_check.head())
print(f"Tổng số hàng: {len(df_check)}")
```
